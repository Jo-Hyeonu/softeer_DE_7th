```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud
```

```python
columns_names = ['target','id', 'data', 'flag', 'user', 'text']
df = pd.read_csv("sentiment.csv", encoding="ISO-8859-1", header=None, names=columns_names)

negative_df = df[df['target'] == 0].sample(frac=1.0, random_state=42)
positive_df = df[df['target'] == 4].sample(frac=1.0, random_state=42)

negative_text = " ".join(negative_df['text'].tolist())
positive_text = " ".join(positive_df['text'].tolist())

fig, axes = plt.subplots(1, 2, figsize=(16, 8))

wc_pos = WordCloud(width=800, height=400, background_color='white', max_words=200, colormap='viridis').generate(positive_text)
axes[0].imshow(wc_pos, interpolation='bilinear')
axes[0].set_title("Positive Sentiment Word Cloud", fontsize=15)
axes[0].axis('off')

wc_neg = WordCloud(width=800, height=400, background_color='black', max_words=200, colormap='Reds').generate(negative_text)
axes[1].imshow(wc_neg, interpolation='bilinear')
axes[1].set_title("Negative Sentiment Word Cloud", fontsize=15)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
